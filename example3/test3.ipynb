```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("hi my name is tomhoon. this is for practice about tokenizer")
```

```python
tokens
```

```python
for token in tokens:
    print(token, encoding.decode([token]))
```
